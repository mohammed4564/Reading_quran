```python
import os

source_data = r"C:\umar_info\Reading_quran\static\pdf\Quran-paras"

for filename in os.listdir(source_data):

    if filename.lower().endswith(".pdf"):

        old_path = os.path.join(source_data, filename)

        # Chapter01.pdf -> 1
        name = os.path.splitext(filename)[0]
        para_number = int(name.replace("Chapter", ""))

        # New filename
        new_filename = f"Holy-Quran-Para-{para_number}.pdf"

        new_path = os.path.join(source_data, new_filename)

        os.rename(old_path, new_path)

        print(f"{filename} -> {new_filename}")
```
